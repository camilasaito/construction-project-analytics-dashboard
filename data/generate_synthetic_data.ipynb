```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import os

# -------------------------------------------------------
# SETTINGS
# -------------------------------------------------------

np.random.seed(42)

OUTPUT_FOLDER = "synthetic_construction_data"
os.makedirs(OUTPUT_FOLDER, exist_ok=True)

today = datetime.today().date()

# -------------------------------------------------------
# PROJECTS
# -------------------------------------------------------

projects = pd.DataFrame({
    "project_id": ["P001", "P002", "P003", "P004", "P005"],
    "project_name": [
        "North Tower",
        "Riverfront Centre",
        "Lakeside Residence",
        "West Industrial Park",
        "Downtown Office"
    ],
    "project_status": [
        "Active",
        "Active",
        "Active",
        "Active",
        "Completed"
    ]
})

projects.to_csv(
    f"{OUTPUT_FOLDER}/projects.csv",
    index=False
)

project_ids = projects["project_id"].tolist()


# -------------------------------------------------------
# HELPER FUNCTIONS
# -------------------------------------------------------

def random_date(days_back=300):
    return today - timedelta(days=int(np.random.randint(1, days_back)))


def create_close_date(created_date, min_days=1, max_days=40):
    return created_date + timedelta(
        days=int(np.random.randint(min_days, max_days))
    )


# -------------------------------------------------------
# RFIs
# Supports:
# Open / Closed
# Late
# Avg Days Late
# Avg Days to Close
# -------------------------------------------------------

rfi_rows = []

for project_id in project_ids:

    for i in range(25):

        created = random_date(300)

        due = created + timedelta(
            days=int(np.random.randint(5, 21))
        )

        status = np.random.choice(
            ["Open", "Closed"],
            p=[0.35, 0.65]
        )

        closed = None

        if status == "Closed":
            closed = create_close_date(created, 2, 40)

        rfi_rows.append({
            "rfi_id": f"{project_id}_RFI_{i+1:03}",
            "project_id": project_id,
            "created_date": created,
            "due_date": due,
            "closed_date": closed,
            "status": status
        })

rfis = pd.DataFrame(rfi_rows)

rfis.to_csv(
    f"{OUTPUT_FOLDER}/rfis.csv",
    index=False
)


# -------------------------------------------------------
# SUBMITTALS
# Supports:
# Open / Closed
# Late
# Avg Days Late
# Avg Days to Close
# -------------------------------------------------------

submittal_rows = []

for project_id in project_ids:

    for i in range(22):

        created = random_date(300)

        due = created + timedelta(
            days=int(np.random.randint(7, 30))
        )

        status = np.random.choice(
            ["Open", "Closed"],
            p=[0.30, 0.70]
        )

        closed = None

        if status == "Closed":
            closed = create_close_date(created, 3, 45)

        submittal_rows.append({
            "submittal_id": f"{project_id}_SUB_{i+1:03}",
            "project_id": project_id,
            "created_date": created,
            "due_date": due,
            "closed_date": closed,
            "status": status
        })

submittals = pd.DataFrame(submittal_rows)

submittals.to_csv(
    f"{OUTPUT_FOLDER}/submittals.csv",
    index=False
)


# -------------------------------------------------------
# OBSERVATIONS
# Supports:
# Open / Closed
# Late
# Avg Days to Close
# -------------------------------------------------------

observation_rows = []

for project_id in project_ids:

    for i in range(20):

        created = random_date(250)

        due = created + timedelta(
            days=int(np.random.randint(5, 25))
        )

        status = np.random.choice(
            ["Open", "Closed"],
            p=[0.25, 0.75]
        )

        closed = None

        if status == "Closed":
            closed = create_close_date(created, 1, 30)

        observation_rows.append({
            "observation_id": f"{project_id}_OBS_{i+1:03}",
            "project_id": project_id,
            "created_date": created,
            "due_date": due,
            "closed_date": closed,
            "status": status
        })

observations = pd.DataFrame(observation_rows)

observations.to_csv(
    f"{OUTPUT_FOLDER}/observations.csv",
    index=False
)


# -------------------------------------------------------
# INSPECTIONS
# Supports:
# Passed
# Action Required
# Days Since Last Inspection
# -------------------------------------------------------

inspection_rows = []

for project_id in project_ids:

    for i in range(18):

        inspection_rows.append({
            "inspection_id": f"{project_id}_INS_{i+1:03}",
            "project_id": project_id,
            "inspection_date": random_date(180),
            "inspection_type": np.random.choice([
                "Quality",
                "Safety",
                "General"
            ]),
            "result": np.random.choice(
                ["Passed", "Action Required"],
                p=[0.78, 0.22]
            )
        })

inspections = pd.DataFrame(inspection_rows)

inspections.to_csv(
    f"{OUTPUT_FOLDER}/inspections.csv",
    index=False
)


# -------------------------------------------------------
# DEFICIENCIES
# Supports:
# Open / Closed
# Priority
# Late
# Avg Days to Close
# -------------------------------------------------------

deficiency_rows = []

for project_id in project_ids:

    for i in range(22):

        created = random_date(250)

        due = created + timedelta(
            days=int(np.random.randint(5, 30))
        )

        status = np.random.choice(
            ["Open", "Closed"],
            p=[0.30, 0.70]
        )

        closed = None

        if status == "Closed":
            closed = create_close_date(created, 1, 35)

        deficiency_rows.append({
            "deficiency_id": f"{project_id}_DEF_{i+1:03}",
            "project_id": project_id,
            "created_date": created,
            "due_date": due,
            "closed_date": closed,
            "status": status,
            "priority": np.random.choice(
                ["Low", "Medium", "High"],
                p=[0.35, 0.45, 0.20]
            )
        })

deficiencies = pd.DataFrame(deficiency_rows)

deficiencies.to_csv(
    f"{OUTPUT_FOLDER}/deficiencies.csv",
    index=False
)


# -------------------------------------------------------
# SAFETY
# Supports:
# Incident types
# Lost Time Injuries
# Recordable Incidents
# Days Since Last LTI
# -------------------------------------------------------

safety_rows = []

for project_id in project_ids:

    # Normal safety events
    for i in range(18):

        incident_type = np.random.choice(
            [
                "Near Miss",
                "First Aid",
                "Medical Treatment",
                "Lost Time Injury"
            ],
            p=[0.40, 0.30, 0.20, 0.10]
        )

        safety_rows.append({
            "safety_id": f"{project_id}_SAF_{i+1:03}",
            "project_id": project_id,
            "incident_date": random_date(365),
            "incident_type": incident_type,
            "lost_time_flag":
                1 if incident_type == "Lost Time Injury" else 0,
            "recordable_flag":
                1 if incident_type in [
                    "Medical Treatment",
                    "Lost Time Injury"
                ] else 0
        })

    # Guarantee at least one LTI record per project
    safety_rows.append({
        "safety_id": f"{project_id}_SAF_LTI",
        "project_id": project_id,
        "incident_date": random_date(300),
        "incident_type": "Lost Time Injury",
        "lost_time_flag": 1,
        "recordable_flag": 1
    })

safety = pd.DataFrame(safety_rows)

safety.to_csv(
    f"{OUTPUT_FOLDER}/safety.csv",
    index=False
)


# -------------------------------------------------------
# SITE ACTIVITY
# Footer:
# People on Site
# Trades on Site
#
# Generates daily records
# -------------------------------------------------------

site_rows = []

for project_id in project_ids:

    for days_ago in range(60):

        activity_date = today - timedelta(days=days_ago)

        site_rows.append({
            "project_id": project_id,
            "activity_date": activity_date,
            "people_on_site": int(
                np.random.randint(25, 150)
            ),
            "trades_on_site": int(
                np.random.randint(3, 18)
            )
        })

site_activity = pd.DataFrame(site_rows)

site_activity.to_csv(
    f"{OUTPUT_FOLDER}/site_activity.csv",
    index=False
)


# -------------------------------------------------------
# DRAWINGS
# Footer:
# Latest Drawing Set
# Latest Revision
# -------------------------------------------------------

drawing_rows = []

drawing_types = [
    "Architectural",
    "Structural",
    "Mechanical",
    "Electrical"
]

for project_id in project_ids:

    for i in range(12):

        drawing_rows.append({
            "drawing_id":
                f"{project_id}_DWG_{i+1:03}",
            "project_id": project_id,
            "drawing_set":
                np.random.choice(drawing_types),
            "revision_number":
                int(np.random.randint(1, 8)),
            "revision_date":
                random_date(180)
        })

drawings = pd.DataFrame(drawing_rows)

drawings.to_csv(
    f"{OUTPUT_FOLDER}/drawings.csv",
    index=False
)


# -------------------------------------------------------
# SUMMARY
# -------------------------------------------------------

print("Synthetic data created successfully!\n")

print("Files created:")
print("- projects.csv")
print("- rfis.csv")
print("- submittals.csv")
print("- observations.csv")
print("- inspections.csv")
print("- deficiencies.csv")
print("- safety.csv")
print("- site_activity.csv")
print("- drawings.csv")

print(f"\nSaved in folder: {OUTPUT_FOLDER}")
```
